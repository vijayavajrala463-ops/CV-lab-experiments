```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
rmin = 100
rmax = 200
img = cv2.imread("/content/drive/MyDrive/mickey-minnie-mouse-mickey-minnie-mouse-playing-each-other-164632863.webp", 0)
withbg = img.copy()
withoutbg = np.zeros(img.shape, dtype=np.uint8)
for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        pv = img[i][j]
        if rmin <= pv <= rmax:
            withbg[i][j] = 255
            withoutbg[i][j] = 255
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("original image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(withoutbg, cmap="gray")
plt.title("Without Background")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(withbg, cmap="gray")
plt.title("With Background")
plt.axis("off")

plt.tight_layout()
plt.show()
```
